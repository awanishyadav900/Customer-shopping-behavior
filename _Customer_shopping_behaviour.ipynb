# finding outliers and remove it

```python
# finding the limits
upper_limit = df['amounts'].mean() + 3*df['amounts'].std()
lower_limit = df['amounts'].mean() - 3*df['amounts'].std()
print('upper_limit :', upper_limit)
print('lower_limit :', lower_limit)
```

```python
# finding the outliers
df.loc[(df['amounts'] > upper_limit ) | (df['amounts']< lower_limit)]
```

```python
# trimming the outliers
new_df = df.loc[(df['amounts'] < upper_limit) & (df['amounts']> lower_limit)]
print('before removing outliers :', len(df))
print('after removing outliers :', len(new_df))
print('outliers :', len(df) - len(new_df))
      
```
